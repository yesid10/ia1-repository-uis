La distribución de Poisson es una distribución de probabilidad que describe el número de eventos que ocurren en un intervalo de tiempo o espacio fijo, bajo la premisa de que estos eventos ocurren con una tasa promedio constante y de manera independiente entre sí. En el contexto del fútbol, se puede usar para modelar la cantidad de goles que un equipo marca en un partido.

La probabilidad de que un equipo marque exactamente \(k\) goles en un partido, si el promedio de goles esperados (\(\lambda\)) es conocido, se calcula usando la fórmula de la distribución de Poisson:

$$
P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}
$$

Donde:
* $P(X = k) \ \ $ es la probabilidad de que el equipo marque $ \ k \ $ goles.
* $\lambda \ \ $ es el promedio de goles que el equipo marca en un partido.
* $k \ \ $ es el número de goles.
* $e \ \ $ es la constante de Euler, aproximadamente igual a 2.71828.
* $k! \ \ $ es el factorial de $k$.


En la práctica, $ \ \lambda \ $ se calcula como el promedio de goles marcados en una serie de partidos. Por ejemplo, si un equipo ha marcado 20 goles en 10 partidos, el valor de $ \ \lambda \ $ sería 2 goles por partido.
