  ```

  outputs = layers.Dense(6, activation='softmax')(x)  # 6 clases
  model = models.Model(inputs=inputs, outputs=outputs)

  optimizer = Adam(learning_rate=0.0005)  # o incluso 0.0005

  model.summary()
  ```
