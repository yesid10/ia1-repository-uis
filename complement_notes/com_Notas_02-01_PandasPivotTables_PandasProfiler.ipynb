# <span style="color:red"> 1. Pivot tables </span>

Pivot tables help to group and summarize dataFrame information, showing information in comapact way. 

For instance, to group and summarize the avergage of houses regarding to the number of ages, we follow next instructions: 

```python
table = pd.pivot_table(d,index=["housing_median_age"],values=["median_house_value"])
table.head(5)
```

We can also use other mathematical functions, by using  the parameter `aggfunc`. For example, in addition to the average value, we want to know the number of houses and the sum of all the values of the houses by age number. Also, the `margins` parameter calculates total numbers at the end of the table. 

```python
table = pd.pivot_table(d,index=["housing_median_age"],values=["median_house_value"],
               aggfunc=[np.mean, np.sum, len], margins=True)
table.head(10)
```
